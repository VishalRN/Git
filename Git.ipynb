```python
l = [2, 3, 5, 8]
for i in l:
    print(i * i,end=" ")
```

```python
for i in range(5,31):
    if i%5==0:
        print(i,end=" ")
```

```python
words = ['python','php','java','json','c']
for i in words:
    if i.startswith('j'):
        print(i,end=" ")
```
